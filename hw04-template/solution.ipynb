# `data_ta`
Contains two sets of data: a profiling/training set, and a set for the attack itself.

1. The profiling_keys.uint8.bin file contains 6,000 encryption keys (6,000 x 16 bytes)
2. The profiling_plaintext.uint8.bin file contains 6,000 plaintexts (6,000 x 16 bytes)
3. The profiling_traces.double.bin file contains 6,000 consumption traces, each consisting of 5,000 samples in double format, i.e. floating point with double precision (6,000 x 5,000 x 8 bytes)
4. The file attack_plaintext.uint8.bin contains 20 plaintexts (20 x 16 bytes)
5. The file attack_traces.double.bin contains 20 consumption traces, each consisting of 5.000 samples in double format, i.e. floating point with double precision (20 x 5.000 x 8 bytes)
6. The data are stored in "binary format", little-endian. This is the format you expect if you are working on a classic PC (Intel, AMD).

```python
from pathlib import Path
import numpy as np

# Profiling:
# 5000 samples [float64]
# 6000 traces

data_dir: Path = Path.cwd() / 'data_ta'

ptrace_length = 5000 # samples
pr_ptraces = np.fromfile(data_dir / 'profiling_traces.double.bin', dtype=np.float64).reshape(-1, ptrace_length)
print(f'Profiling power traces shape: {pr_ptraces.shape} [{pr_ptraces.dtype}]')

pt_length = 16 # Bytes
pr_plaintexts = np.fromfile(data_dir / 'profiling_plaintext.uint8.bin', dtype=np.uint8).reshape(-1, pt_length)
print(f'Profiling plaintexts shape: {pr_plaintexts.shape} [{pr_plaintexts.dtype}]')

key_length = 16 # Bytes
pr_keys = np.fromfile(data_dir / 'profiling_keys.uint8.bin', dtype=np.uint8).reshape(-1, key_length)
print(f'Profiling keys shape: {pr_keys.shape} [{pr_keys.dtype}]')

assert pr_ptraces.shape[0] == pr_plaintexts.shape[0] == pr_keys.shape[0]

# trace length same as profiled traces
at_ptraces = np.fromfile(data_dir / 'attack_traces.double.bin').reshape(-1, ptrace_length)
print(f'Attack power traces shape: {at_ptraces.shape} [{at_ptraces.dtype}]')

at_plaintexts = np.fromfile(data_dir / 'attack_plaintext.uint8.bin', dtype=np.uint8).reshape(-1, pt_length)
print(f'Attack plaintexts traces shape: {at_plaintexts.shape} [{at_plaintexts.dtype}]')
```

```python
def init_list_of_lists(n_lists):
    return [ [] for _ in range(n_lists) ]
        
def calc_intermediate_value(val, key_guess):
    return SBox[val ^ key_guess]

def calc_leakage_function(val):
    return bin(val).count('1') # Hamming Weight

def partition_traces(pr_ptraces, pr_plaintexts, pr_keys, key_byte_idx):
    n_partitions = 9
    partitions = init_list_of_lists(n_partitions)

    for pr_trace, pr_pt, pr_key in zip(pr_ptraces, pr_plaintexts, pr_keys):
        v = calc_intermediate_value(pr_pt[key_byte_idx], pr_key[key_byte_idx])
        L = calc_leakage_function(v) # L (leakage value) is an integer between [0, 8]
        assert L < n_partitions
        partitions[L].append(pr_trace)

    return partitions

def calc_mean_traces(partitioning):
    mean_traces = [] # n_partitions-count (9) mean traces
    for i, partition in enumerate(partitioning):
        mean_traces.append(
            np.mean(np.stack(partition), axis=0) # the average trace (columns squished together)
        )
    return mean_traces

def select_poi(mean_traces, n_pois: int):
    # calculate sum of differences for the partitioned traces
    sod = np.zeros(ptrace_length) # sum of differences trace (ptrace_length long)
    for i, m_trace_i in enumerate(mean_traces):
        for j, m_trace_j in enumerate(mean_traces):
            
            if i != j:
                sod = sod + np.abs(m_trace_i - m_trace_j)
    # select n points of interest with largest sum of differences
    return np.flip(np.argsort(sod))[0:n_pois]
    
def comp_gauss_params(partitioning, mean_traces, pois):
    # select points of interest only
    restricted_mean_traces = [ m_trace[pois] for m_trace in mean_traces ]
    
    # noises = [
    #   [noise trace of each trace in partition_0], # each len(noises[i][j]) == len(pois)
    #   ...
    #   [noise trace of each trace in partition_N]
    # ]
    noises = init_list_of_lists(len(partitioning))
    assert len(partitioning) == len(restricted_mean_traces)

    for partition, m_trace, part_noise in zip(partitioning, restricted_mean_traces, noises):
        for ptrace in partition:
            part_noise.append(ptrace[pois] - m_trace)
    
    gauss_params = []
    # compute noise covariances between all points of interest
    for noise_partition in noises:
        gauss_params.append( 
            np.cov(
                np.stack(noise_partition).T
            )
        )
    
    # each gaussian parameters is a symmetric noise covariance matrix
    return gauss_params
```

# Profiling phase

```python
partitioning = partition_traces(pr_ptraces, pr_plaintexts, pr_keys, key_byte_idx=0)

mean_traces = calc_mean_traces(partitioning)
pois = select_poi(mean_traces, n_pois=5)
gauss_params = comp_gauss_params(partitioning, mean_traces, pois)
    
templates = dict()
for i in range(9):
    templates[i] = (mean_traces[i][pois], gauss_params[i])
```

# Attack phase

```python
from scipy.stats import multivariate_normal

def calc_matching_p_scipy(trace, template):
    """Alternative: Calculate the log-likelihood using scipy's built-in multivariate_normal."""
    mean_vec = template[0]
    cov_mtx = template[1]
    
    mvn = multivariate_normal(mean=mean_vec, cov=cov_mtx)
    
    return mvn.logpdf(trace)

def calc_matching_p(trace, template):
    """Calculate probability the trace matches the template."""
    n_pois = len(trace)
    
    mean_vec = template[0]
    cov_mtx = template[1]

    inv_cov_mtx = np.linalg.inv(cov_mtx)
    det_cov_mtx = np.linalg.det(cov_mtx)

    nom   = np.exp( - 1/2 * (trace - mean_vec) @ inv_cov_mtx @ (trace - mean_vec).T )
    denom = np.sqrt( ( (2 * np.pi) **  n_pois) * det_cov_mtx )
    return np.log( nom / denom )
```

```python
def attack(matching_f, n_traces: int):
    P_sums = np.zeros(256)
    at_pts = at_plaintexts[0:n_traces]
    at_ptrcs = at_ptraces[0:n_traces]

    for key_guess in range(256):
        for at_pt, at_trace in zip(at_pts, at_ptrcs):
            L = calc_leakage_function(calc_intermediate_value(at_pt[0], key_guess))
            template = templates[L]
            P = matching_f(at_trace[pois], template)
            P_sums[key_guess] += P

    return P_sums, np.argmax(P_sums)
```

```python
real_key = 0x2b

for n_traces in range(1, at_plaintexts.size):
    P_sums_scipy, key_scipy = attack(calc_matching_p_scipy, n_traces)
    P_sums, key             = attack(calc_matching_p, n_traces)
    
    if key == real_key:
        print(f'Found key: {hex(key)} using {n_traces} attack traces [my matching func]')
    
    if key_scipy == real_key:
        print(f'Found key: {hex(key_scipy)} using {n_traces} attack traces [scipy matching func]')
    
    if key_scipy == key == real_key:
        break
```
