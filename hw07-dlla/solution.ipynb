```python
from pathlib import Path
import numpy as np

data_dir: Path = Path.cwd()

ptrace_length = 4250 # unmasked samples per waveform

c_ptraces_um = np.fromfile(data_dir / 'nomask' / 'constant-traces-nomask.bin', dtype=np.int16).reshape(-1, ptrace_length)
print(c_ptraces_um.shape)

r_ptraces_um = np.fromfile(data_dir / 'nomask' / 'random-traces-nomask.bin', dtype=np.int16).reshape(-1, ptrace_length)

print(r_ptraces_um.shape)

ptrace_length = 8875 # masked samples per waveform

c_ptraces_m = np.fromfile(data_dir / 'mask' / 'constant-traces-mask.bin', dtype=np.int16).reshape(-1, ptrace_length)
print(c_ptraces_m.shape)

r_ptraces_m = np.fromfile(data_dir / 'mask' / 'random-traces-mask.bin', dtype=np.int16).reshape(-1, ptrace_length)

print(r_ptraces_m.shape)
```

```python
# non-specific student's t-test
t = (np.mean(c_ptraces_um, axis=0) - np.mean(r_ptraces_um, axis=0)) / np.sqrt(np.var(c_ptraces_um, axis=0) / c_ptraces_um.shape[0] + np.var(r_ptraces_um, axis=0) / r_ptraces_um.shape[0])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(t)
plt.axhline(4.5, color='red', linestyle='-')
plt.axhline(-4.5, color='red', linestyle='-')
```
