# 1. Download the prepared data
The power traces were measured during PRESENT encryption implemented in an FPGA with an 80-bit key. 
- Each power sample is stored as a signed 16-bit integer (also known as int16_t, signed short).

- The file constant-traces-nomask.bin contains 10,000 power traces measured with constant fixed plaintexts. Each power trace consists of 4.250 samples.

- The file random-traces-nomask.bin contains 10.000 power traces measured with random plaintexts. Each power trace consists of 4.250 samples.

- The file random-plaintexts-nomask.bin is not required, we have simplified the assignment in the end.

- The file constant-traces-mask.bin contains 10.000 power traces measured with constant fixed plaintexts. Each power trace consists of 8.875 samples.

- The file random-traces-mask.bin contains 10.000 power traces measured with random plaintexts. Each power trace consists of 8.875 samples.

- The data are stored in "binary format", little-endian. This is the format you expect if you are working on a classic PC (Intel, AMD).

- In every file, first, there is a respective number of samples collected during the first encryption. After that, there are samples collected during the second encryption. And so on.

```python
from pathlib import Path
import numpy as np

data_dir: Path = Path.cwd() / 'ni-hsc-lab-10-data'
nm_ptrace_length = 4250 # samples in each non-masked power trace

c_ptraces_nm = np.fromfile(data_dir / 'nomask' / 'constant-traces-nomask.bin', dtype=np.int16).reshape(-1, nm_ptrace_length)
print(f'Constant power traces of non-masked implementation shape: {c_ptraces_nm.shape}')

r_ptraces_nm = np.fromfile(data_dir / 'nomask' / 'random-traces-nomask.bin', dtype=np.int16).reshape(-1, nm_ptrace_length)
print(f'Random power traces of non-masked implementation shape: {r_ptraces_nm.shape}')

m_ptrace_length = 8875 # samples in each masked power trace

c_ptraces_m = np.fromfile(data_dir / 'mask' / 'constant-traces-mask.bin', dtype=np.int16).reshape(-1, m_ptrace_length)
print(f'Constant power traces of non-masked implementation shape: {c_ptraces_m.shape}')

r_ptraces_m = np.fromfile(data_dir / 'mask' / 'random-traces-mask.bin', dtype=np.int16).reshape(-1, m_ptrace_length)
print(f'Random power traces of non-masked implementation shape: {r_ptraces_m.shape}')
```
